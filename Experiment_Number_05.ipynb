## Experiment: Reasoning Techniques (Forward Chaining)

### Simple Code: Understanding Forward Chaining

Forward chaining is a reasoning technique that starts with known facts and applies inference rules to derive new conclusions. It moves forward from the known data to the desired goal. Think of it as answering the question: "What can I conclude from what I know?"

This simple example demonstrates how a system can deduce whether someone is prepared for rain based on two initial facts: if it's raining and if they have an umbrella.

```python
# Define a set of known facts (initial knowledge base)
facts = {
    'is_raining': True, # Fact: It is raining
    'has_umbrella': True # Fact: The person has an umbrella
}

# Define a set of rules (if-then statements)
rules = [
    # Rule 1: If it's raining AND the person has an umbrella, then they are prepared for rain
    {'condition': ['is_raining', 'has_umbrella'], 'consequence': 'is_prepared_for_rain'},
    # Rule 2: If it's raining AND the person does NOT have an umbrella, then they will get wet
    {'condition': ['is_raining', 'not has_umbrella'], 'consequence': 'will_get_wet'}
]

# Initialize a set to store newly inferred facts
inferred_facts = set()

# Function to perform forward chaining
def forward_chaining(current_facts, inference_rules):
    new_facts_inferred = True # Flag to check if any new facts were inferred in an iteration
    while new_facts_inferred:
        new_facts_inferred = False # Reset flag for the current iteration
        # Iterate through each rule
        for rule in inference_rules:
            conditions_met = True # Assume all conditions of the rule are met initially
            # Check each condition in the rule
            for condition in rule['condition']:
                is_negated = condition.startswith('not ') # Check if the condition is negated
                actual_condition = condition[4:] if is_negated else condition # Extract the actual condition name

                # If it's a negated condition (e.g., 'not has_umbrella')
                if is_negated:
                    # Condition is met if the actual_condition is NOT in current_facts
                    if actual_condition in current_facts:
                        conditions_met = False
                        break
                # If it's a positive condition (e.g., 'is_raining')
                else:
                    # Condition is met if the actual_condition IS in current_facts
                    if actual_condition not in current_facts:
                        conditions_met = False
                        break

            # If all conditions for the rule are met
            if conditions_met:
                consequence = rule['consequence'] # Get the consequence of the rule
                # If this consequence is not already in our known facts
                if consequence not in current_facts:
                    current_facts[consequence] = True # Add the new fact to our knowledge base
                    inferred_facts.add(consequence) # Add to the list of newly inferred facts
                    new_facts_inferred = True # Set flag to true, indicating new facts were found
    return current_facts # Return the updated set of facts

# Convert initial facts to a dictionary for easier lookup in the forward chaining function
initial_facts_dict = {fact: True for fact, value in facts.items() if value}

# Run forward chaining
final_facts = forward_chaining(initial_facts_dict, rules)

# Display the initial facts
print("Initial Facts:")
for fact, value in facts.items():
    print(f"- {fact}: {value}")

print("\n--- Running Forward Chaining ---")

# Display the inferred facts
print("\nInferred Facts:")
if inferred_facts:
    for fact in inferred_facts:
        print(f"- {fact}: True")
else:
    print("No new facts were inferred.")

# Display the final state of all known facts
print("\nFinal Knowledge Base:")
for fact, value in final_facts.items():
    print(f"- {fact}: {value}")
```

### Explanation of the Simple Code:

*   **`facts = {...}`**: This dictionary stores our initial set of known information. For example, `is_raining: True` means "it is raining." This is our starting knowledge base.
*   **`rules = [...]`**: This list contains our inference rules. Each rule is a dictionary with:
    *   `'condition'`: A list of facts that must be true (or not true) for the rule to fire.
    *   `'consequence'`: The new fact that becomes true if all conditions are met.
    *   Notice `not has_umbrella` – this indicates a negated condition.
*   **`inferred_facts = set()`**: This set will store all the new facts that the forward chaining process discovers.
*   **`def forward_chaining(current_facts, inference_rules):`**: This function implements the core forward chaining algorithm.
*   **`new_facts_inferred = True`**: A boolean flag to control the main loop. The loop continues as long as new facts are being inferred.
*   **`while new_facts_inferred:`**: This loop ensures that the system keeps applying rules until no more new facts can be derived from the current knowledge base.
*   **`for rule in inference_rules:`**: Each rule is checked against the current facts.
*   **`conditions_met = True`**: This flag tracks whether all conditions of the current rule have been satisfied.
*   **`for condition in rule['condition']:`**: Each condition within a rule is evaluated.
*   **`is_negated = condition.startswith('not ')`**: Checks if a condition is a negative one (e.g., "not has umbrella").
*   **`actual_condition = ...`**: Extracts the actual fact name from the condition (e.g., 'has_umbrella' from 'not has_umbrella').
*   **`if is_negated:`**: If the condition is negated, it's met if the `actual_condition` is *not* in `current_facts`.
*   **`else:`**: If the condition is positive, it's met if the `actual_condition` *is* in `current_facts`.
*   **`if conditions_met:`**: If all conditions of a rule are found to be true in the `current_facts`:
    *   **`consequence = rule['consequence']`**: The new fact that the rule concludes is identified.
    *   **`if consequence not in current_facts:`**: This crucial check prevents redundant additions and ensures the loop terminates. If the consequence is already known, it's not a *new* fact.
    *   **`current_facts[consequence] = True`**: The newly inferred fact is added to our knowledge base.
    *   **`inferred_facts.add(consequence)`**: The new fact is also added to our `inferred_facts` set.
    *   **`new_facts_inferred = True`**: The flag is set to `True` so the `while` loop continues for another iteration to see if this new fact can trigger other rules.
*   **`return current_facts`**: After the loop finishes (no new facts can be inferred), the function returns the complete set of facts.
*   The print statements at the end display the initial facts, the newly inferred facts, and the final state of the knowledge base.

### Complex Application: Simple Medical Diagnosis using Forward Chaining

This example extends the concept of forward chaining to a more practical scenario: a simplified medical diagnosis system. Given a set of symptoms, the system will use a predefined set of rules to deduce potential conditions or recommendations. This illustrates how AI can be used in expert systems for decision support.

```python
# Define a more extensive set of initial symptoms (facts)
patient_symptoms = {
    'has_fever': True,        # Patient has a fever
    'has_cough': True,        # Patient has a cough
    'has_sore_throat': False, # Patient does not have a sore throat
    'has_headache': True,     # Patient has a headache
    'is_fatigued': True       # Patient is fatigued
}

# Define a more complex set of diagnostic and recommendation rules
diagnosis_rules = [
    # Rule 1: If fever and cough, then suspect flu
    {'condition': ['has_fever', 'has_cough'], 'consequence': 'suspect_flu'},
    # Rule 2: If sore throat and no fever, then suspect cold
    {'condition': ['has_sore_throat', 'not has_fever'], 'consequence': 'suspect_cold'},
    # Rule 3: If suspect flu and headache, then strongly suspect flu
    {'condition': ['suspect_flu', 'has_headache'], 'consequence': 'strongly_suspect_flu'},
    # Rule 4: If strongly suspect flu and fatigued, then recommend rest
    {'condition': ['strongly_suspect_flu', 'is_fatigued'], 'consequence': 'recommend_rest'},
    # Rule 5: If suspect cold, then recommend over-the-counter_meds
    {'condition': ['suspect_cold'], 'consequence': 'recommend_otc_meds'},
    # Rule 6: If suspect flu, then advise to consult doctor
    {'condition': ['suspect_flu'], 'consequence': 'advise_consult_doctor'},
    # Rule 7: If not has_fever and not has_cough and not has_sore_throat, then likely_no_illness
    {'condition': ['not has_fever', 'not has_cough', 'not has_sore_throat'], 'consequence': 'likely_no_serious_illness'}
]

# Initialize a set to store newly inferred facts for this application
inferred_diagnoses = set()

# Function to perform forward chaining (re-using the same logic)
def forward_chaining_diagnosis(current_facts, inference_rules, inferred_store):
    new_facts_inferred = True
    while new_facts_inferred:
        new_facts_inferred = False
        for rule in inference_rules:
            conditions_met = True
            for condition in rule['condition']:
                is_negated = condition.startswith('not ')
                actual_condition = condition[4:] if is_negated else condition

                if is_negated:
                    if actual_condition in current_facts:
                        conditions_met = False
                        break
                else:
                    if actual_condition not in current_facts:
                        conditions_met = False
                        break

            if conditions_met:
                consequence = rule['consequence']
                if consequence not in current_facts:
                    current_facts[consequence] = True
                    inferred_store.add(consequence) # Store newly inferred fact in the dedicated set
                    new_facts_inferred = True
    return current_facts

# Convert initial symptoms to a dictionary for easier lookup
initial_symptoms_dict = {symptom: True for symptom, value in patient_symptoms.items() if value}

# Run forward chaining for diagnosis
final_diagnosis_facts = forward_chaining_diagnosis(initial_symptoms_dict, diagnosis_rules, inferred_diagnoses)

# Display the initial symptoms
print("Patient's Initial Symptoms:")
for symptom, value in patient_symptoms.items():
    print(f"- {symptom}: {value}")

print("\n--- Running Diagnosis System (Forward Chaining) ---")

# Display the inferred diagnoses and recommendations
print("\nInferred Diagnoses and Recommendations:")
if inferred_diagnoses:
    for fact in inferred_diagnoses:
        print(f"- {fact}: True")
else:
    print("No specific diagnoses or recommendations inferred from symptoms.")

# Display the final state of all known facts including symptoms and diagnoses
print("\nFinal Knowledge Base (Symptoms + Inferences):")
for fact, value in final_diagnosis_facts.items():
    print(f"- {fact}: {value}")
```

### Explanation of the Complex Application Code:

*   **`patient_symptoms = {...}`**: This dictionary represents the initial input, which are the symptoms reported by a patient. This is the starting point of our knowledge for the diagnosis.
*   **`diagnosis_rules = [...]`**: This list contains the diagnostic logic. Each rule attempts to connect symptoms to potential conditions (`suspect_flu`, `suspect_cold`) or further actions (`recommend_rest`, `advise_consult_doctor`). The rules can also chain together, meaning one inferred fact (like `suspect_flu`) can be a condition for another rule.
    *   For example, Rule 1 (`'condition': ['has_fever', 'has_cough'], 'consequence': 'suspect_flu'`) suggests that if a patient has both a fever and a cough, then flu is suspected.
    *   Rule 3 (`'condition': ['suspect_flu', 'has_headache'], 'consequence': 'strongly_suspect_flu'`) shows how an already inferred fact (`suspect_flu`) can combine with another symptom (`has_headache`) to lead to a stronger conclusion.
    *   Rules can also lead to recommendations (e.g., `recommend_rest`, `advise_consult_doctor`).
    *   Rule 7 includes multiple `not` conditions, demonstrating that the absence of symptoms can also lead to a conclusion.
*   **`inferred_diagnoses = set()`**: This set specifically tracks the diagnoses and recommendations that the system infers, separating them from the initial symptoms.
*   **`def forward_chaining_diagnosis(...)`**: This is essentially the same forward chaining function as in the simple example, but it takes an additional argument `inferred_store` to store the newly inferred facts specific to this application.
*   **`initial_symptoms_dict = {...}`**: Converts the initial `patient_symptoms` dictionary into a format suitable for the `forward_chaining_diagnosis` function.
*   **`final_diagnosis_facts = forward_chaining_diagnosis(...)`**: Executes the forward chaining process, applying the `diagnosis_rules` to the `patient_symptoms`.
*   The final print statements output:
    *   The initial symptoms provided by the patient.
    *   The specific diagnoses and recommendations derived by the system through forward chaining.
    *   The complete knowledge base at the end, including both the original symptoms and all inferred facts.

## Experiment: Reasoning Techniques (Backward Chaining)

### Simple Code: Understanding Backward Chaining

This simple example uses backward chaining to determine if a student should 'study for exam' based on their current 'knowledge level' and if the 'exam is tomorrow'. The system starts with the goal and tries to find facts or rules that support it.

```python
# Define a set of known facts (initial knowledge base)
facts_bc = {
    'has_low_knowledge': True, # Fact: Student has low knowledge
    'exam_is_tomorrow': True   # Fact: Exam is tomorrow
}

# Define a set of rules (if-then statements) for backward chaining
rules_bc = [
    # Rule 1: If student has low knowledge AND exam is tomorrow, then they should study for exam
    {'consequence': 'should_study_for_exam', 'conditions': ['has_low_knowledge', 'exam_is_tomorrow']},
    # Rule 2: If student is procrastinating, then they have low knowledge
    {'consequence': 'has_low_knowledge', 'conditions': ['is_procrastinating']},
    # Rule 3: If student has good grades, then they do not have low knowledge
    {'consequence': 'not has_low_knowledge', 'conditions': ['has_good_grades']}
]

# Function to perform backward chaining
def backward_chaining(goal, current_facts, inference_rules, visited_goals=None):
    # Initialize visited_goals set to prevent infinite loops in recursive calls
    if visited_goals is None:
        visited_goals = set()

    # If the goal is already in our facts, it's proven
    if goal in current_facts:
        return True

    # Handle negated goals (e.g., 'not has_low_knowledge')
    is_negated_goal = goal.startswith('not ')
    actual_goal = goal[4:] if is_negated_goal else goal

    # If the actual goal is directly stated as false or its negation is true
    if actual_goal in current_facts and not current_facts[actual_goal] and not is_negated_goal:
        return False
    if actual_goal in current_facts and current_facts[actual_goal] and is_negated_goal:
        return False

    # Prevent infinite recursion for cyclical rules
    if goal in visited_goals:
        return False
    visited_goals.add(goal)

    # Try to find a rule that concludes the goal
    for rule in inference_rules:
        consequence = rule['consequence']

        # Check if this rule's consequence matches our current goal (or its negation)
        if (consequence == goal) or \n           (is_negated_goal and consequence == actual_goal and rule.get('is_negating_rule', False)) or \n           (not is_negated_goal and consequence == f'not {goal}' and rule.get('is_negating_rule', False)):
            all_conditions_met = True
            # For each condition in the rule, try to prove it recursively
            for condition in rule['conditions']:
                # If a sub-goal cannot be proven, then this rule does not lead to our goal
                if not backward_chaining(condition, current_facts, inference_rules, visited_goals.copy()):
                    all_conditions_met = False
                    break
            # If all conditions for this rule are met, then the goal is proven
            if all_conditions_met:
                # If the goal is a positive fact, add it to current_facts
                if not is_negated_goal:
                    current_facts[goal] = True
                # If the goal is a negated fact, add the positive form as False to current_facts
                else:
                    current_facts[actual_goal] = False
                return True # Goal proven

        # Special handling for 'not' consequences in rules (like Rule 3)
        elif consequence.startswith('not ') and consequence[4:] == actual_goal:
            negated_consequence_actual = consequence[4:]
            all_conditions_met = True
            for condition in rule['conditions']:
                if not backward_chaining(condition, current_facts, inference_rules, visited_goals.copy()):
                    all_conditions_met = False
                    break
            if all_conditions_met:
                current_facts[negated_consequence_actual] = False # Explicitly state the fact is false
                if is_negated_goal: # If our goal was 'not has_low_knowledge', then it's proven
                    return True

    return False # Goal cannot be proven with current facts and rules

# The goal we want to prove
goal_bc = 'should_study_for_exam'

print(f"Goal: {goal_bc}")
print(f"Initial Facts: {facts_bc}")

# Run backward chaining to prove the goal
is_proven = backward_chaining(goal_bc, facts_bc, rules_bc)

print(f"\nCan '{goal_bc}' be proven? {is_proven}")

if is_proven:
    print("Reasoning path (inferred/used facts):")
    for fact, value in facts_bc.items():
        if value and fact != goal_bc:
            print(f"- {fact}: {value}")
    print(f"- {goal_bc}: True (concluded)")
else:
    print(f"The goal '{goal_bc}' cannot be established with the given facts and rules.")
```

### Explanation of the Simple Code:

*   **`facts_bc = {...}`**: This dictionary holds our initial known facts about the student.
*   **`rules_bc = [...]`**: This list defines our knowledge base of rules. Each rule has a `'consequence'` (the conclusion it proves) and a list of `'conditions'` (what needs to be true for the conclusion).
    *   Notice Rule 3: `{'consequence': 'not has_low_knowledge', 'conditions': ['has_good_grades']}`. This rule allows us to infer the *negation* of a fact, which is important in many reasoning systems.
*   **`def backward_chaining(...)`**: This is the core function implementing the backward chaining algorithm.
*   **`goal`**: The specific fact or conclusion we are trying to prove.
*   **`current_facts`**: The current knowledge base, including initial facts and any facts inferred during the process.
*   **`inference_rules`**: The set of rules to use for inference.
*   **`visited_goals`**: A set used to keep track of goals already visited in the current path of recursion. This prevents infinite loops if there are circular dependencies in the rules (e.g., A -> B, B -> A).
*   **`if goal in current_facts:`**: If our goal is already directly present in our knowledge base, it's proven immediately.
*   **`is_negated_goal = goal.startswith('not ')`**: Handles goals that are negations (e.g., 'not has_low_knowledge').
*   **Preventing Infinite Recursion**: `if goal in visited_goals:` checks if we're trying to prove a goal that is already in our current reasoning path, avoiding endless loops.
*   **`for rule in inference_rules:`**: The system iterates through rules to find one whose `consequence` matches the `goal`.
*   **Matching Consequences**: The condition `if (consequence == goal) or ...` checks if the rule's consequence directly matches the goal, or if it's a rule that negates a fact that is related to the current (possibly negated) goal.
*   **`for condition in rule['conditions']:`**: If a rule's consequence matches the goal, each of its conditions becomes a *sub-goal* that the `backward_chaining` function tries to prove recursively.
*   **`if not backward_chaining(condition, ...):`**: If any sub-goal cannot be proven, then this particular rule cannot be used to prove the main goal, and the system tries the next rule.
*   **`if all_conditions_met:`**: If all conditions of a rule are successfully proven (recursively), then the original `goal` is considered proven, and `True` is returned.
    *   **Adding Inferred Facts**: If the goal is proven, it's added to `current_facts` for future reference, either as `True` for positive facts or `False` for negated facts.
*   **`return False`**: If no rule can prove the goal and it's not in the initial facts, the function returns `False`.
*   The print statements at the end show whether the goal was proven and list the facts that led to its conclusion.

### Complex Application: Basic Troubleshooting System using Backward Chaining

This example simulates a simple troubleshooting system using backward chaining. Let's say our goal is to determine 'why the internet is not working'. The system will work backward from this goal, checking various possible causes based on a set of rules and current observations.

```python
# Define initial observations (facts)
troubleshooting_facts = {
    'router_lights_off': True, # Observation: Router lights are off
    'computer_connected_wifi': True, # Observation: Computer is connected to WiFi
    'modem_working': True # Observation: Modem is working fine
}

# Define troubleshooting rules (consequence -> conditions)
troubleshooting_rules = [
    # Goal: internet_not_working
    {'consequence': 'internet_not_working', 'conditions': ['router_issue']},
    {'consequence': 'internet_not_working', 'conditions': ['modem_issue']},
    {'consequence': 'internet_not_working', 'conditions': ['isp_down']},

    # Sub-goal: router_issue
    {'consequence': 'router_issue', 'conditions': ['router_lights_off']},
    {'consequence': 'router_issue', 'conditions': ['router_unplugged']},

    # Sub-goal: modem_issue
    {'consequence': 'modem_issue', 'conditions': ['not modem_working']},

    # Sub-goal: isp_down (requires user input in a real system, here we pre-assume for demonstration)
    {'consequence': 'isp_down', 'conditions': ['provider_outage_reported']},

    # Sub-goal: router_unplugged (to demonstrate deeper nesting, not in initial facts)
    {'consequence': 'router_unplugged', 'conditions': ['power_cord_loose']}
]

# Function for backward chaining (re-using the logic from the simple example)
def backward_chaining_troubleshoot(goal, current_facts, inference_rules, visited_goals=None):
    if visited_goals is None:
        visited_goals = set()

    if goal in current_facts:
        # Check if the fact's value is explicitly False for a negated goal
        if goal.startswith('not ') and current_facts.get(goal[4:]) is False: # e.g., if goal is 'not modem_working' and 'modem_working' is False
            return True
        # Check if the fact's value is explicitly True for a positive goal
        if not goal.startswith('not ') and current_facts.get(goal) is True: # e.g., if goal is 'router_issue' and 'router_issue' is True
            return True
        # If it's a 'not' goal and the positive version is True, then the 'not' goal is False
        if goal.startswith('not ') and current_facts.get(goal[4:]) is True:
            return False
        # If it's a positive goal and the positive version is False, then the goal is False
        if not goal.startswith('not ') and current_facts.get(goal) is False:
            return False

    is_negated_goal = goal.startswith('not ')
    actual_goal = goal[4:] if is_negated_goal else goal

    # Prevent infinite recursion
    if goal in visited_goals:
        return False
    visited_goals.add(goal)

    for rule in inference_rules:
        consequence = rule['consequence']

        # Check if this rule's consequence directly proves our current goal
        if consequence == goal:
            all_conditions_met = True
            for condition in rule['conditions']:
                if not backward_chaining_troubleshoot(condition, current_facts, inference_rules, visited_goals.copy()):
                    all_conditions_met = False
                    break
            if all_conditions_met:
                # If goal is positive, add it as True. If goal is negative, add actual_goal as False.
                current_facts[actual_goal] = (not is_negated_goal)
                return True
        # Handle rules where the consequence is a negation (e.g., 'not modem_working')
        elif consequence.startswith('not ') and consequence[4:] == actual_goal:
             if is_negated_goal: # If our goal is also a negation, and the rule negates the same fact
                all_conditions_met = True
                for condition in rule['conditions']:
                    if not backward_chaining_troubleshoot(condition, current_facts, inference_rules, visited_goals.copy()):
                        all_conditions_met = False
                        break
                if all_conditions_met:
                    current_facts[actual_goal] = False # Explicitly state the fact is false
                    return True # Goal ('not actual_goal') proven

    return False

# The goal we want to troubleshoot
goal_troubleshoot = 'internet_not_working'

print(f"Troubleshooting Goal: {goal_troubleshoot}")
print(f"Initial Observations: {troubleshooting_facts}")

# Run backward chaining
is_internet_down = backward_chaining_troubleshoot(goal_troubleshoot, troubleshooting_facts, troubleshooting_rules)

print(f"\nCan '{goal_troubleshoot}' be confirmed? {is_internet_down}")

if is_internet_down:
    print("\nPossible cause(s) identified:")
    # Iterate through the rules to find which ones led to the final conclusion
    for rule in troubleshooting_rules:
        if rule['consequence'] == 'internet_not_working':
            # Check if any of its conditions were inferred as true
            for condition in rule['conditions']:
                actual_condition = condition[4:] if condition.startswith('not ') else condition
                if actual_condition in troubleshooting_facts and troubleshooting_facts[actual_condition] == (not condition.startswith('not ')):
                    print(f"- {condition}")
                    # This is a simplification; a full traceback would be more complex

    print("\nFinal state of identified issues/facts:")
    for fact, value in troubleshooting_facts.items():
        if fact != goal_troubleshoot and value is True:
            print(f"- {fact}: {value}")
    if troubleshooting_facts.get(goal_troubleshoot, False):
        print(f"- {goal_troubleshoot}: True (concluded)")

else:
    print(f"\n'{goal_troubleshoot}' cannot be confirmed with the given observations and rules. Further investigation needed.")
```

### Explanation of the Complex Application Code:

*   **`troubleshooting_facts = {...}`**: This dictionary stores the initial observations or symptoms reported by the user when trying to troubleshoot an internet issue.
*   **`troubleshooting_rules = [...]`**: This list contains the diagnostic rules for the troubleshooting system. Each rule states a `consequence` (a potential problem or intermediate state) and the `conditions` that must be true for that consequence to be valid.
    *   Rules are structured hierarchically: `internet_not_working` can be caused by `router_issue`, `modem_issue`, or `isp_down`. Each of these in turn has its own conditions.
    *   For example, `router_issue` is a sub-goal, and it can be proven if `router_lights_off` is true.
    *   The rule `{'consequence': 'modem_issue', 'conditions': ['not modem_working']}` demonstrates how the absence of a fact (`modem_working` being `False`) can lead to a conclusion.
*   **`def backward_chaining_troubleshoot(...)`**: This function is a refined version of the `backward_chaining` function, adapted for the troubleshooting context.
    *   **Improved Fact Checking**: It explicitly checks for `True` or `False` values in `current_facts` for both positive and negated goals.
    *   The recursive logic remains similar: when a rule's consequence matches the current `goal`, it tries to prove each of the rule's `conditions` as sub-goals.
    *   If a rule's consequence is itself a negation (e.g., `'not modem_working'`) and the current goal is also that negation, it attempts to prove the rule's conditions.
*   **`goal_troubleshoot = 'internet_not_working'`**: This is the primary problem we are trying to diagnose.
*   The print statements at the end display:
    *   The initial observations.
    *   Whether the `internet_not_working` goal was confirmed.
    *   If confirmed, it attempts to list the specific conditions (or issues) that led to the conclusion. This is a simplified output for explanation; a real system might provide a more detailed trace of the reasoning path.
