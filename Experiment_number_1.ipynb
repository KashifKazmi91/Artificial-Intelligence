**Numpy**

**What is NumPy?**
NumPy (Numerical Python) is the fundamental Python library for numerical computing. It provides a powerful N-dimensional array (ndarray) object and a collection of mathematical functions to perform fast operations on arrays.
Almost every AI and Machine Learning library (such as Pandas, Scikit-learn, TensorFlow, and PyTorch) is built on top of NumPy.

**Why is NumPy used in AI?**
Efficient storage of numerical data; Fast mathematical computations;  Matrix operations for Machine Learning; Data preprocessing; Linear Algebra operations; Random number generation; Foundation for Deep Learning

**Real-World AI Applications of NumPy**
Storing image pixels as arrays; Processing sensor data from robots; Representing datasets for machine learning; Performing matrix multiplication in neural networks; Data normalization before model training

```python
import numpy as np
# Step 1: Import the NumPy library
import numpy as np

# Step 2: Create a NumPy array
numbers = np.array([10, 20, 30, 40, 50])

# Step 3: Print the array
print("Array:", numbers)

# Step 4: Find the sum of all elements
print("Sum =", np.sum(numbers))

# Step 5: Find the average of the elements
print("Average =", np.mean(numbers))
```

**Pandas**

**What is Pandas?**
Pandas is an open-source Python library used for data manipulation, analysis, and preprocessing. It helps organize data into tables (similar to Excel sheets or SQL tables), making it easier to clean, filter, analyze, and prepare data for AI and Machine Learning models.

The name Pandas comes from Panel Data, referring to multidimensional structured data.

**Why is Pandas used in AI?**
Before building any AI or Machine Learning model, we need to:
Load datasets; Clean missing values; Filter unwanted data; Analyze data; Select important features; Prepare data for training

These tasks are efficiently handled using Pandas.

```python
# Step 1: Import the Pandas library
import pandas as pd

# Step 2: Create a dictionary containing student data
data = {
    "Name": ["Amit", "Neha", "Rahul"],
    "Marks": [85, 90, 78]
}

# Step 3: Convert the dictionary into a DataFrame
df = pd.DataFrame(data)

# Step 4: Display the DataFrame
print(df)
```

**Matplotlib**

**What is Matplotlib?**
Matplotlib is one of the most popular Python libraries used for data visualization. It helps convert numerical data into graphs and charts, making it easier to understand trends, patterns, and relationships.

In AI and Machine Learning, visualization is an essential step because it allows us to explore datasets before training models and interpret results after training.

**Why is Matplotlib used in AI?**
Matplotlib is commonly used to:

Visualize datasets; Compare different categories; Identify trends and patterns; Detect outliers; Display model performance; Plot training and validation accuracy/loss

```python
# Step 1: Import pyplot module from matplotlib
import matplotlib.pyplot as plt

# Step 2: Create student names
students = ["Amit", "Neha", "Rahul"]

# Step 3: Create corresponding marks
marks = [85, 90, 78]

# Step 4: Draw the bar chart
plt.bar(students, marks)

# Step 5: Give title to the graph
plt.title("Student Marks")

# Step 6: Label X-axis
plt.xlabel("Students")

# Step 7: Label Y-axis
plt.ylabel("Marks")

# Step 8: Display the graph
plt.show()
```

**Scipy**

**What is SciPy?**
SciPy (Scientific Python) is an open-source Python library used for scientific computing, mathematics, optimization, statistics, and signal processing. It is built on top of NumPy, which means SciPy uses NumPy arrays as its basic data structure.

In AI and Machine Learning, SciPy is used whenever we need advanced mathematical operations, statistical analysis, optimization, or distance calculations.

**Why is SciPy used in AI?**

SciPy helps AI engineers to:
Perform statistical analysis; Optimize machine learning models; Calculate distances between data points; Process signals and images; Solve mathematical equations; Perform scientific computations

```python
# Step 1: Import the statistics module from SciPy
from scipy import stats

# Step 2: Create a list of marks
marks = [70, 75, 80, 85, 90]

# Step 3: Calculate the mean
average = stats.tmean(marks)

# Step 4: Calculate the median
middle = stats.median_abs_deviation(marks)

# Step 5: Display the results
print("Average Marks:", average)
print("Median Absolute Deviation:", middle)
```

**Sci-kit Learn**

**What is Scikit-learn?**

Scikit-learn (commonly written as sklearn) is one of the most popular Python libraries used for Machine Learning. It provides ready-made algorithms for classification, regression, clustering, model evaluation, and data preprocessing.
It is built on top of NumPy, Pandas, and SciPy, making it an essential library for AI and Machine Learning.

**Why is Scikit-learn used in AI?**

Scikit-learn helps us:

Build Machine Learning models; Train models using datasets; Make predictions; Evaluate model performance; Preprocess data; Split datasets into training and testing sets

```python
# Step 1: Import Linear Regression model
from sklearn.linear_model import LinearRegression

# Step 2: Training data (Hours Studied)
X = [[1], [2], [3], [4], [5]]

# Step 3: Output data (Marks)
y = [35, 45, 55, 65, 75]

# Step 4: Create the model
model = LinearRegression()

# Step 5: Train the model
model.fit(X, y)

# Step 6: Predict marks for 6 study hours
prediction = model.predict([[6]])

# Step 7: Display the prediction
print("Predicted Marks:", prediction[0])
```

**Tensorflow**

**What is TensorFlow?**

TensorFlow is an open-source Deep Learning framework developed by Google. It is used to build, train, and deploy Artificial Intelligence (AI) and Deep Learning models.
Unlike Scikit-learn, which is mainly used for traditional Machine Learning algorithms, TensorFlow is designed to solve complex AI problems involving large datasets and neural networks.

Why is TensorFlow used in AI?

TensorFlow is used to:

Build Artificial Neural Networks (ANN)
Develop Deep Learning models; Image Recognition; Speech Recognition; Natural Language Processing (NLP); Object Detection; Recommendation Systems; Generative AI (LLMs)

```python
# Step 1: Import TensorFlow
import tensorflow as tf

# Step 2: Create a tensor
numbers = tf.constant([10, 20, 30, 40])

# Step 3: Display the tensor
print(numbers)
```

**Keras**

**What is Keras?**

Keras is a high-level Deep Learning API used to build and train Artificial Neural Networks (ANNs). It provides a simple and user-friendly interface for creating deep learning models.
Originally, Keras was developed as an independent library by François Chollet. Today, it is integrated into TensorFlow as tf.keras, making it the recommended way to build neural networks in TensorFlow.

**Simple Definition:**
Keras is a high-level API that makes it easy to build, train, and test deep learning models using TensorFlow.

**Why is Keras used in AI?**

Keras is used to:

Build Artificial Neural Networks (ANN); Develop Deep Learning models; Perform Image Classification; Speech Recognition; Natural Language Processing (NLP); Predict future values; Build Recommendation Systems

```python
# Step 1: Import Sequential model and Dense layer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Step 2: Create the model
model = Sequential()

# Step 3: Add one neuron to the model
model.add(Dense(1))

# Step 4: Display the model structure
model.summary()
```

**Pytorch**

**What is PyTorch?**

PyTorch is an open-source Deep Learning framework developed by Meta AI (Facebook AI Research). It is widely used for building and training Artificial Neural Networks (ANNs), especially in research and advanced AI applications.

Like TensorFlow, PyTorch can build deep learning models, but it is known for its simplicity, flexibility, and dynamic computation graph.

**Why is PyTorch used in AI?**

PyTorch is used for:

Deep Learning; Artificial Neural Networks; Computer Vision; Natural Language; Processing (NLP); Speech Recognition; Robotics; Generative AI; Large Language Models (LLMs)

```python
# Step 1: Import PyTorch library
import torch

# Step 2: Create a tensor
numbers = torch.tensor([10, 20, 30, 40])

# Step 3: Display the tensor
print("Tensor:", numbers)

# Step 4: Add 5 to every element
result = numbers + 5

# Step 5: Display the updated tensor
print("Updated Tensor:", result)
```

**Seaborn**

**What is Seaborn?**

Seaborn is a Python library used for creating statistical data visualizations. It is built on top of Matplotlib, which means it uses Matplotlib internally but provides more attractive, informative, and easier-to-create graphs.

In AI and Machine Learning, Seaborn is widely used for Exploratory Data Analysis (EDA) before training models.

**Why is Seaborn used in AI?**

Seaborn helps AI engineers to:

Understand datasets visually; Identify patterns and relationships; Detect outliers; Analyze feature distributions; Find correlations between variables; Visualize machine learning results

```python
# Step 1: Import Seaborn library
import seaborn as sns

# Step 2: Import Matplotlib for displaying the graph
import matplotlib.pyplot as plt

# Step 3: Create student names
students = ["Amit", "Neha", "Rahul"]

# Step 4: Create marks
marks = [85, 90, 78]

# Step 5: Draw the bar plot
sns.barplot(x=students, y=marks)

# Step 6: Give title
plt.title("Student Marks")

# Step 7: Display graph
plt.show()
```
